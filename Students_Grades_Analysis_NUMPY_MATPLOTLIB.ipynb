******Student Grades Analysis.******

 In this project, we will use NumPy to analyze students' grades in different subjects.

```python
import numpy as np
```

***Making the Dataset.***

```python
grades = np.array([
    [85, 90, 78],
    [70, 75, 80],
    [92, 88, 95],
    [60, 65, 70],
    [78, 82, 85]
    ])
grades
```

***Checking the Shape of the Data.***

We check the shape to understand how many students and subjects are included in our dataset.

```python
grades.shape
```

So we have 5 students and 3 subjects.

***Calculating the Average Grade for Each Student.***

We calculate the average grade for each student to understand their overall performance.

```python
student_average = np.mean(grades, axis = 1)
student_average = np.round(student_average, 2)

student_average.reshape(5,1)
```

***Finding the Highest and Lowest Student Average.***

```python
highest_average = np.max(student_average)
lowest_average = np.min(student_average)

print("Highest Average:", highest_average)
print("Lowest Average:", lowest_average)
```

***Calculating the Average Grade for Each Subject.***

```python
subject_average = np.mean(grades, axis = 0, keepdims=True)
subject_average = np.round(subject_average, 2)
subject_average
```

***Finding the Highest and Lowest Performing Subject***

```python
highest_subject_index = np.argmax(subject_average)
lowest_subject_index = np.argmin(subject_average)
print("Highest Subject Average Index:", highest_subject_index)
print("Lowest Subject Average Index:", lowest_subject_index)
```

***Assigning Subject Names.***

We assign names to the three columns to make the analysis easier to understand.

```python
subjects = np.array(["Math", "Programming", "English"])
subjects
```

```python
for i in range(len(subjects)):
 print(subjects[i], "Average is:", subject_average[0][i])
```

***Identifying the Highest and Lowest Performing Students.***

```python
best_student_index = np.argmax(student_average)
worst_student_index = np.argmin(student_average)

print("Best Student Index:", best_student_index)
print("Average of Best Student:", student_average[best_student_index])
print("Worst Student Index:", worst_student_index)
print("Average of Worst Student:", student_average[worst_student_index])
```

******Visualizing Average Grades by Subject.******

A bar chart is used to compare the average grade across the three subjects

```python
import matplotlib.pyplot as plt

plt.bar(subjects, subject_average[0], color=['b', 'g', 'r'])
plt.title("Average Grades by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Grade")
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--')
plt.show()
```

******Visualizing Student Average Grades.******

A bar chart is used to compare the overall average grade of each student.

```python
plt.bar(["Student 1", "Student 2", "Student 3", "Student 4", "Student 5"], student_average, color=['b','g','r','y','m'])
plt.title("Average Grade for Each Student")
plt.xlabel("Student")
plt.ylabel("Average Grade")
plt.show()
```

- The lowest-performing student was (Student 4), with an average of (65.0) .

- The subject with the lowest average was (Math) .
